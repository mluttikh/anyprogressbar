# Getting started

A progress bar says how far a loop is and about how long it will still take. Wrap what you
loop over in `progress(...)`. In a notebook the bar is a widget (install
`anyprogressbar[widget]`), or plain HTML without it; the same code in a terminal draws a
line, and in a log file writes one every few seconds.

```python
import time

from anyprogressbar import progress
```

```python
for row in progress(range(60), "Loading", unit="rows"):
    time.sleep(0.05)
```

The bar gives no time remaining at first: it waits until the estimate is worth showing.
When the loop ends the bar stays as a record of what happened, and that record is saved
with the notebook.

Work that is done within half a second shows nothing at all:

```python
for row in progress(range(1000), "Too quick to show"):
    pass
```

## By hand

Not every job is a loop over a list. Give a total and report as you go. In a `with` block
the bar ends when the block does.

```python
sizes = [120, 40, 300, 80, 260, 200]

with progress(total=sum(sizes), desc="Downloading", unit="MB") as bar:
    for size in sizes:
        time.sleep(size / 300)
        bar.update(size)
```

## Numbers beside the bar

`set_postfix` puts values of your own after the bar, and `set_description` changes its name.

```python
with progress(total=40, desc="Training", unit="steps") as bar:
    for step in range(40):
        time.sleep(0.08)
        bar.update()
        bar.set_postfix(loss=1 / (step + 1), lr=0.001)
```

## Without a total

When you do not know how much there is, the bar claims no percentage: a block runs along
the track, and the count and the rate say how the work goes.

```python
def rows():
    for i in range(50):
        time.sleep(0.05)
        yield i

for row in progress(rows(), "Scanning", unit="rows"):
    pass
```

If you do know the size of something that has no length, pass `total=`:

```python
for row in progress(rows(), "Scanning", total=50, unit="rows"):
    pass
```
