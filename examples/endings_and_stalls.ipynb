# How a bar ends, and what it says when nothing happens

A bar ends in one of three ways, and says which: **done** (green), **failed** (red, with
the reason) or **stopped** (amber: it ended before the work was finished). The words are
there as well as the colour.

```python
import time

from anyprogressbar import progress
```

## Failed

An exception inside a `with` block ends the bar as failed, with the exception as the
reason. The exception itself goes on as usual; here it is caught so that the rest of the
notebook runs.

```python
try:
    with progress(total=10, desc="Importing", unit="files") as bar:
        for i in range(10):
            time.sleep(0.15)
            if i == 6:
                raise ValueError("row 7 has no date")
            bar.update()
except ValueError:
    pass
```

You can also say it yourself:

```python
bar = progress(total=5, desc="Uploading").start()
time.sleep(0.6)
bar.update(2)
bar.fail("the server said no")
```

## Stopped

A loop that is left early ends its bar as stopped. So does a `with` block that ends before
the total is reached.

```python
for i in progress(range(100), "Searching"):
    time.sleep(0.02)
    if i == 40:
        break
```

An exception in the body of a loop ends the loop's bar as failed too:

```python
try:
    for i in progress(range(30), "Parsing", unit="lines"):
        time.sleep(0.05)
        if i == 20:
            int("twenty")
except ValueError:
    pass
```

## A slow item

The widget's clock runs in the browser, so the bar keeps time while one item takes long.
(The plain HTML bar only changes when the loop reports.) After ten seconds without
progress, and five times what an item took until then, the bar says so instead of giving a
time remaining that is no longer true.

```python
with progress(total=6, desc="Mostly quick", unit="files") as bar:
    for i in range(6):
        time.sleep(12 if i == 3 else 0.5)
        bar.update()
```
